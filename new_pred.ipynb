## Checking trained logistic regression model on new data

```python
import numpy as np
import pandas as pd
import joblib
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from pathlib import Path
```

```python
DATA_PATH = Path('data')/'test.csv'

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f'Data File not Found {DATA_PATH.resolve()}'
    )

df = pd.read_csv(DATA_PATH)
```

```python
df.head()
```

```python
df = df.drop(columns=['Unnamed: 0','id'])
```

```python
df.head()
```

```python
df.info()
```

```python
df['satisfaction'] = df['satisfaction'].map({
    'neutral or dissatisfied':0,
    'satisfied':1
})
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['satisfaction'])
y = df['satisfaction']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,stratify=y,random_state=40)
```

```python
trained_model = joblib.load('logistic-reg_trained')
```

```python
ypred_trained = trained_model.predict(X_train)
```

```python
ypred_test = trained_model.predict(X_test)
```

```python
print(f'accuracy score of X train data - {accuracy_score(y_train,ypred_trained)}')
```

```python
print(f'accuracy score of X train data - {accuracy_score(y_test,ypred_test)}')
```

```python
print('Classification Report of X train data')
print(classification_report(y_train,ypred_trained))
```

```python
print('Classification Report of X test data')
print(classification_report(y_test,ypred_test))
```
